```python
import pandas as pd
```

```python
#load dataset
df = pd.read_csv('Housing.csv')
```

```python
#criteria 
BUDGET_MIN   = 2000000   # 20 lakh
BUDGET_MAX   = 5000000   # 50 lakh
MIN_BEDROOMS = 3
FURNISHING   = "furnished"   # "furnished" / "semi-furnished" / "unfurnished"
NEED_BASEMENT = "yes"        # "yes" / "no"
```

```python
#filter
result = df[
    (df["price"]            >= BUDGET_MIN) &
    (df["price"]            <= BUDGET_MAX) &
    (df["bedrooms"]         >= MIN_BEDROOMS) &
    (df["furnishingstatus"] == FURNISHING) &
    (df["basement"]         == NEED_BASEMENT)]
```

```python
#sorting
result = result.sort_values(
    by=["area", "stories"],
    ascending=[False, False]
)
```

```python
#Result 
print(f"\n {len(result)} ghar mile tumhare criteria par!\n")
print(result.to_string(index=False))
```
